```python
a=3
b=3

if a > b:
    print("a is greater than b")
elif a < b:
    print("a is less than b")
else:
    print("a is equal to b")


a=2.0
b=2.1
if a<b or a>b or a!=b:
    print("a is not equal to b")
else:
    print("a is equal to b")
```

```python
def evenT_or_odd(n):
    return True if n % 2 == 0 else "odd"
print(evenT_or_odd(4))
print(evenT_or_odd(7))
```

```python
name = "CS0"
match name:
    case "CS50":
        print("Matched CS50")
    case "CS500":
        print("Matched CS500")
    case _: # _ means any value that doesn't match previous cases
        print("No match")
```
