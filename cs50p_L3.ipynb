```python
# emample of return and using break
def find_first_even(numbers):
    for number in numbers:
        if number % 2 == 0:
            return number
    return None

numbers = [1, 3, 5, 6, 8]
i=find_first_even(numbers)
print(i)

# Using break instead of return

for number in numbers:
    if number % 2 == 0:
        print(number)
        break  
else:
    print("No even number found")  

```

```python
def main():
    a= aik_simple_func("kuch bi enter karo: ")
    print(a)

def aik_simple_func(prompt):
    while True:
        try:
            return (int(input(prompt)))*5   # the return will break out of the loop and return the result also
        except ValueError:
            pass        # pass does nothing, just ignores invalid input and prompts again, loop continues

main()
```
