```python
import random
print(random.choice(["tail", "head"]))
print(random.randint(1, 10))  # prints a random integer between 1 and 10 inclusive
```

```python
from random import choice, randint
print(choice(["tail", "head"]))
print(randint(1, 10))  
```

```python
import random

cards = ["jack", "queen", "king"]
random. shuffle(cards)
print(cards)
```

```python
import statistics

data = [1, 2, 3, 4, 4, 5]
print(statistics.mean(data))   # mean is the average of the data set
print(statistics.median(data)) # median mean that the middle value of the data set
print(statistics.mode(data))   # mode is the most frequently occurring value in the data set
```
