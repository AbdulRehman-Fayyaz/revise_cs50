# What is pytest and how to use it
Pytest is a testing framework for Python that allows you to write test cases for your code to ensure it behaves as expected. It is widely used for unit testing and can be easily integrated into your development workflow.
To use pytest, you typically follow these steps:

1. Install pytest using pip:
```bash                                 
pip install pytest
```

2. Create a test file, usually named `test_<module>.py`, where `<module>` is the name of the module you want to test.

3. Write test functions within the test file. Each test function should start with `test_` and contain assertions to check the expected behavior of your code.

4. Run pytest from the command line:
```bash
pytest file_name.py
```

Pytest will automatically discover and run all test functions in files matching the pattern `test_*.py` and report the results.
